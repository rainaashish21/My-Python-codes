```python
#22
x = [("Apple", 500), ("Banana", 200), ("Mango", 900), ("Guava", 750)]

total = 0
expensive = x[0]
g = int(input("Enter a value: "))
abv = []

for item in x:
    total += item[1]

    if item[1] > expensive[1]:
        expensive = item

    if item[1] > g:
        abv.append(item[0])

print(f"Total cost is {total}")
print(f"Most expensive item is {expensive[0]}")
print(f"Above {g} items are {abv}")
```

```python
#23
L=[]
Salary={"Ashish":10000,"Ram":30000, "Sham":40000}
total_salary = 0
for values in Salary.values():
  total_salary =values+total_salary
avg_salary =total_salary/len(Salary)
print("Average Salary is ",avg_salary)

for value in Salary.values():
  L.append(value)
print(f"Maximum salary is {max(L)}")
print(f"Minimum Salary is {min(L)}")
print(f"Average Salary is {avg_salary}")

for key,value in Salary.items():
  if value>avg_salary:
    print(key,value)
  else:
    print(key,"Salary is less tha average")
```

```python
#24
Available_books ={"Alpha","Beta","Gamma"}
Issued_Books =[]

while True:
  print("1.Available books\n2.Add_books\n3.ISSUE Books\4.return book\n5.Exit")
  n=int(input("Enter the option: "))
  if n==1:
    print(Available_books)
  elif n==2:
    add_book = input("Enter book name to enter")
    Available_books.add(add_book)
    print(Available_books)
  elif n==3:
    issue = input("Enter issue book : ")
    Issued_Books.append(issue)
    Available_books.remove(issue)
    print(Available_books)
    print(Issued_Books)
  elif n==4:
    returned = input("ENter return book : ")
    Issued_Books.remove(returned)
    Issued_Books.append(returned)
  elif n==5:
    print("Thank You")
    break
  else:
    print("Enter valid option")
```
