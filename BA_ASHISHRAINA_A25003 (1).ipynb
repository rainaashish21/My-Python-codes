```python
#24. Check Armstrong number.
num = int(input("Enter the number:"))
n= num
power = len(str(num))
total =0
while n>0:
  digit = n%10
  total = total+digit ** power
  n= n//10
if total == num:
  print(f"The {num} is Armstrong Number")
else:
  print(f"The {num} is  not Armstrong Number")
```

```python
#25. Print all Armstrong numbers between 1 to 1000.
def armstrong(n):
  temp = n
  power = len(str(n))
  total =0
  while n>0:
    digit = n%10
    total = total+digit ** power
    n= n//10
  return total==temp
for i in range(1,1001):
    if armstrong(i):
      print(i)

```
