```python
def bill_amount(a,b):
  if b<=100:
    c= a+(b*5)
    return c
  elif b<=200:
    d= a + (100*5)+((b-100)*7)
    return d
  elif b>200:
    e= a+ (100*5) +(100*7)+((b-200)*10)
    return e

x = 100
y= int(input("Enter the unit consumed:"))
bill_amount= bill_amount(x,y)
print(f"Rupees is {bill_amount}")
```

```python
# Simple billing system
def Tot_cost(a,b):
  c=a*b
  return c

x= int(input("Enter the quantity consumed:"))
y=int(input("Enter the price of that quantity"))
Cost = Tot_cost(x,y)
print(Cost)
if Cost>1000:
    print(f"Discount amount is {0.1*Cost}")
else:
    print("No discount")
Final_cost= Cost-(0.1*Cost)
print(f"Final cost after discount is {Final_cost}")
```

```python
balance = 10000

while True:
  x= int(input("Choose option:\n 1.Check Balance\n 2.Deposit \n 3. Withdraw \n 4.Exit\n"))
  if x==1:
    print("Balance amount is:",balance)
  elif x==2:
    DP= int(input("Enter the amount to deposit:"))
    if DP>0:
      balance = balance + DP
      print("The balance amount is:",balance)
    else:
      print("Invalid amount")
  elif x==3:
    W= int(input("Enter the amount to withdraw:"))
    if W <=balance and W>0:
      balance= balance - W
      print("The balance amount is:",balance)
    else:
      print("Invalid amount")
  elif x==4:
    print("thank You")
    break
  else:
    print("Invalid option")
```
